```python
import yfinance as yf 
```

```python
tikers = ["AAPL", "NVDA", "MSFT"]
datos = yf.download(tikers, period="1y")

precios = datos['Close']
```

# Graficos interactivos

```python
import plotly.express as px
```

```python
grafica = px.line(precios, x=precios.index, y=precios.columns, title="Precios de Cierre de AAPL, NVDA y MSFT en el ultimo año",
                  template = "plotly_dark", labels = {"value": "Precio (USD)", "variable": "Acción"})
```

```python
grafica 
```

```python
px.line(precios["AAPL"],
        title = "Cotizacion historica de AAPL(ultimos doce meses)",
        labels = {"Value": "Precio (USD)", "Date": "Fecha"},
        template = "plotly_dark")
```

```python
px.bar(precios.mean(axis=0),
       title = "Promedio de precios de cierre de AAPL, NVDA Y MSFT en el ultimo año",
       labels = {"value": "Precio (USD)", "index": "Accion"},
       color = precios.mean(axis=0),
       template = "plotly_dark",)
```

## Acceder a las filas de nuestros datos

```python
fig.write_html("inversion.html")
```
